1. What is NumPy, and why is it widely used in Python?


NumPy is a Python library used for numerical computing. It provides fast operations on large arrays and matrices.


It’s widely used because it’s efficient, supports mathematical operations, and integrates well with other libraries.


2. How does broadcasting work in NumPy?



Broadcasting allows NumPy to perform operations on arrays of different shapes. It automatically expands smaller arrays to match larger ones.

3. What is a Pandas DataFrame?


A DataFrame is a table-like structure in Pandas with rows and columns, like an Excel sheet.


It’s used for handling and analyzing structured data.


4. Explain the use of the groupby() method in Pandas.


```python
# groupby() groups data based on a column, allowing you to perform operations (e.g., sum, mean) on each group.
#Example:

import pandas as pd
data = pd.DataFrame({'Category': ['A', 'B', 'A'], 'Value': [10, 20, 30]})
print(data.groupby('Category').sum())

```

Python lists: Flexible types, slower, no built-in math operations.


7. What is a heatmap, and when should it be used?

8. What does the term “vectorized operation” mean in NumPy?


```python
#Vectorized operations perform computations on entire arrays without loops, making them faster.

#Example:

import numpy as np
a = np.array([1, 2, 3])
print(a * 2)  # Output: [2, 4, 6]
```

1. How do you create a 2D NumPy array and calculate the sum of each row?


```python
#Create a 2D array using np.array and use the np.sum function with axis=1 to sum rows.

import numpy as np
array = np.array([[1, 2, 3], [4, 5, 6]])
row_sums = np.sum(array, axis=1)
print(row_sums)  # Output: [6, 15]
```

2. Write a Pandas script to find the mean of a specific column in a DataFrame.


```python
#Use DataFrame['column_name'].mean() to calculate the mean.
import pandas as pd
data = pd.DataFrame({'Scores': [80, 90, 100]})
mean_score = data['Scores'].mean()
print(mean_score)  # Output: 90.0
```

3. Create a scatter plot using Matplotlib.


```python
#Use plt.scatter(x, y) to create a scatter plot.

import matplotlib.pyplot as plt
x = [1, 2, 3]
y = [4, 5, 6]
plt.scatter(x, y)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Scatter Plot')
plt.show()
```

4. How do you calculate the correlation matrix using Seaborn and visualize it with a heatmap?


```python
#Use df.corr() to calculate the correlation matrix and sns.heatmap to visualize it.


import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
data = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})
correlation_matrix = data.corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.show()

```

5. Generate a bar plot using Plotly.


```python
#Use plotly.express.bar() to create a bar plot.

import plotly.express as px
data = {'Categories': ['A', 'B', 'C'], 'Values': [10, 20, 30]}
fig = px.bar(data, x='Categories', y='Values')
fig.show()

```

6. Create a DataFrame and add a new column based on an existing column.

7. Write a program to perform element-wise multiplication of two NumPy arrays.


```python
#Use the * operator for element-wise multiplication.

import numpy as np
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
result = a * b
print(result)  # Output: [4, 10, 18]
```

8. Create a line plot with multiple lines using Matplotlib.


```python
#Use plt.plot() for multiple lines.

import matplotlib.pyplot as plt
x = [1, 2, 3]
y1 = [2, 4, 6]
y2 = [1, 3, 5]
plt.plot(x, y1, label='Line 1')
plt.plot(x, y2, label='Line 2')
plt.legend()
plt.show()
```

9. Generate a Pandas DataFrame and filter rows where a column value is greater than a threshold.


```python

import pandas as pd
data = pd.DataFrame({'Scores': [70, 80, 90]})
filtered_data = data[data['Scores'] > 75]
print(filtered_data)
```

10. Create a histogram using Seaborn to visualize a distribution.



```python
#Use sns.histplot() to create a histogram.

import seaborn as sns
import matplotlib.pyplot as plt
data = [1, 2, 2, 3, 3, 3, 4, 4, 5]
sns.histplot(data, bins=5, kde=True)
plt.show()
```

11. Perform matrix multiplication using NumPy.



```python
#Use np.dot() or @ for matrix multiplication.

import numpy as np
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
result = np.dot(a, b)
print(result)
```
